## Basic Aggregation

```python
# Sum of all values in the entire array
import numpy as np
# 100 days of sales data across 4 categories
# Rows = days  Columns = Categories
np.random.seed(6)
daily_sales0 = np.random.normal(loc=13000, scale= 1500, size= (10,4)).round(0).astype(int)
print(daily_sales0.round(0))
growth_rates = np.array([1.02,1.06,0.98,2.34])
compound = growth_rates.prod()
print(f"Compound Effect: {compound:.4f}")

#Cumulative sum
food_cumsum = daily_sales0[:,0].cumsum()
print(f"Running Food Revenue:,{food_cumsum}")

# Cumulative Product   (To show the growth rate )
rates = np.array([1.0,1.02,1.05,0.98,1.03])
print(f"Compound after each period: {rates.cumprod().round(4)}")
```

# Statistical Aggregation


```python
food_sales = daily_sales0[:,0]
# Mean
print(f"Mean Food Sales: {food_sales.mean():.0f}")
# Median
print(f"Median Food Sales: {np.median(food_sales):.0f}")
# Standard Dev
print(f"Standard Dev: {food_sales.std():.0f}")
# Variance
print(f"Variance: {food_sales.var():.0f}")
# Highest
print(f"Highest Day: {food_sales.max():.0f}")
# Lowest
print(f"Lowest Day: {food_sales.min():.0f}")
```

# Boolean Indexing
- Boolean Indexing lets you select elements based on a condition. Numpy evals the condition, creates a True/False array(a mask), and returns only the elements where the condition is True

```python
food_sales = daily_sales0[:,0]
mask = food_sales > 10000
print(mask)
# We then can use the mask to get the actual values
high_food_sales = food_sales[mask]
print(high_food_sales)

# This gives the number of days that exceeded the target
print((food_sales > 13000).sum())
# The same can Be done to get the mean
```

# Multiple Conditions

```python
# We use AND(&) OR(|) , NOT(~)
good_days = daily_sales0[(daily_sales0[:,0] > 13000) & (daily_sales0[:,1] > 10000)]
print(good_days)

concern_days = daily_sales0[(daily_sales0[:,0] > 13000) | (daily_sales0[:,1] > 10000)]
print(concern_days)
```

# NP.WHERE: Conditional Logic on Arrays

```python
#Syntax np.where(condition,value if true, value if false)
#Example:
# Was each day's food 
food_sales = daily_sales0[:,0]
food_label = np.where(food_sales > 13000,"High", "Low")
high = food_sales[food_label == "High"]    # Gets only those values that are in the high category
print(high)   
print(food_label)
# Return Numeric Values
food_flag = np.where(food_sales > 13000, 1, 0)
print(food_flag)
# Keep Values above threshhold, replace others with the threshold
floor_13k = np.where(food_sales < 13000, 13000, food_sales)
print(floor_13k)
```

# Handling missing Data(NaN)
- It is a special float value that means "there is no valid number here" - It is the absence of a value
- Any arithmetic NaN returns nAn which is why you have to handle it explicity

```python
import numpy as np
# NaN is a Float
print(type(np.nan))

# Arithmetic with NaN will ALWAYS return NaN
print(5+np.nan)

# NaN is never equal to itself
print(np.nan == np.nan)

# The right way to check for NaN
print(np.isnan(np.nan))
```

```python
# Inserting and Detecting NaN 
# Simulate a sales dataset with missing values
sales = np.array([[12000,45000,np.nan,22000,40000],
                  [np.nan,20000,35000,9500,12500],
                  [11000,np.nan,20300,12000,71340],
                  [15000,4500,9000,23000,np.nan]
                  ])
# Detecting NaN
nan_mask = np.isnan(sales[:,0])
print(nan_mask)

# Count the missing values
print(f"Total Missing Values: {nan_mask.sum()}")

# To print the missing values per row or column, add axis= in the sum() : sum(axis=1)
```

## Replace Specific Values with NaN


```python
raw = np.array([19,30,-1,30,20])
clean = np.where(raw ==-1, np.nan, raw)
print(clean)
```

# Removing NaN using np.isnan()

```python
food_with_gaps = np.array([19,30,-1,30,20])
gaps = np.where(food_with_gaps ==-1, np.nan,food_with_gaps)

valid_only = gaps[(~np.isnan(gaps))]
print(valid_only)
```
