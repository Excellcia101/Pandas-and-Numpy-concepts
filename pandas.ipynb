### Series and Dataframes

#### Series
Is a 1-D array that can hold any data type

```python
import pandas as pd
#Create a series
sr = pd.Series([10,20,30,40])
print(sr)
sr1 = pd.Series([10,20,30,40], index= ["a","b","c","d"])
print(sr1)
```

#### Dataframe
It is a 2-D labelled data structure with columns that can be of different types

```python
import pandas as pd
data = {"Name": ["Alice","Bob","Charity","Danson"],
        "Age": [20,19,21,20],
        "City": ["Berlin","Tokyo","Shibuya","Washington"]
        }
df = pd.DataFrame(data, index= ["A","B","C","D"])
print(df)
```

```python
import pandas as pd

# Create a series using range()
# range(10, 50, 10) generates: 10, 20, 30, 40 (stops before 50)
sr = pd.Series(range(10, 50, 10)) 
print("Series with range data:")
print(sr)

# Create a series with range data and custom index
sr1 = pd.Series(range(10, 50, 10), index=["a", "b", "c", "d"])
print("\nSeries with range data and custom index:")
print(sr1)
```

## **Filtering Rows**
Filtering rows in Pandas means selecting only the rows that meet specific conditions.
This is done by creating a boolean mask (a list of True/False values) that tells Pandas which rows to keep
#### *Ways of Filtering rows*
1. Boolean Indexing
2. Using (.query()) 

## Boolean Indexing
**Syntax**
df[condition]

Example: filtered = df[df['Age'] > 30]  #[df['Age'] > 30] is the condition

To fit in multiple conditions we use the following:

AND (&) (filtered = df[(df['City'] == 'NY') & (df['Age'] >= 25)]) Finds people in NY with the age of 25+

OR(|)  (filtered = df[(df['City'] == 'NY') | (df['City'] == 'PAR')]) Finds people either in NY or PAR

NOT(~)  filtered = df[~(df['City'] == 'NY')] Finds people not in NY

If you have a list of values to match, don't write multiple | operators. Use .isin().
*Example:    
*             cities = ['NY', 'PAR', 'TKY']
*            filtered = df[df['City'].isin(cities)]
notna() is used to check missing values in a particular column.
Keeps only rows where 'City' is NOT missing
filtered = df[df['City'].notna()]



## Using .query()
The .query() method allows you to write conditions as strings. It makes it more readable.
Example:

*           filtered = df.query("Age > 30 and City == 'NY' or City == 'PAR'")
#### Note: You can use 'and'/'or' instead of & and | inside the string!

In lists:

*    min_age = 30
*    target_cities = ['NY', 'PAR']

filtered = df.query("Age > @min_age and City in @target_cities")

### Grouping And Aggregation
#### Groupby
- it works in 3 steps:
    - Split - Divides the data into groups
    -Apply - runs a function on each group
    -Combine - Brings results together
- "for Each [category], calculate[something]"

```python
df.groupby('job_title') ["salary_kes"].mean()
df.groupby('city')['name'].count()
df.groupby('employment_type')['salary_kes'].max()
```

###agg() - Multiple Statistics at once

```python
df.groupby('job_title')['salary_kes'].agg(['mean','median','min','max','count'])
df.groupby('job_title').agg({
            'salary_kes': ['mean','max'],
            'experience_yrs': ['mean','min'],
            'age': 'median'})
```

# Grouping by Multiple Columns

```python
df.groupby(['city''employment_type'])['salary_kes'].mean()

df.groupby(['city''employment_type']).agg({
    'salary_kes': 'mean',
    'experience_yrs': 'mean'})
```

### Pivot Tables
These do the same thing as groupby


```python
pd.pivot_table(
    df,
    values= "salary_kes",
    index= "job_title",
    columns= "employment_type",
    aggfunc= "mean"
    margins= True
    margins_name= "Overall Avg"
    fill_value=0   # This helps fill in nan values in the set with 0 to prevent the code from crashing
)
```

## Dropping Missing Values

df.dropna(how='all')    # Drops only if all values in the row are missing 

df.dropna(subset= ['salary_history_2022']) #Drops rows if missing in specific columns

df.dropna(axis=1)   # Drops columns with any missing values

df.dropna(thresh=8)   # Keeps rows with at least 8
